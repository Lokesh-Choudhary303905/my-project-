```python
student_name = str(input("enter the student name: "))
student_class = int(input("enter the student class: "))
subjects=['English','math','hindi','chemistry','physics']
marks = []
sum=0
for i in range(5) :
    #subjects=input("enter the subject name:")
    marks.append(int(input("enter the student marks in " + subjects[i] + ":")))
    sum=sum+marks[i]
for i in range(0,5):
    print(marks[i])
print("student name: ",student_name)
print("student class: ",student_class)
print("total marks= ",sum)
print("percentage",sum/5)
```

```python
str1=input('enter the first string:')
str2=input('enter the first string:')
str3= (str1 +str2)
print(str3)
print(str3.lower())
print(str3.upper())
print(str3.title())
print(str3.swapcase())
print(str3.capitalize())
print(str3.casefold())
print(str3.center(30,'*'))
print(str3.count('a'))
print(str3.endswith('h'))
print(str3.find('k'))
print(str3.isalnum())
print(str3.isdigit())
print(str3.isnumeric())
print(str3.isspace())
print(str3.replace('o','a'))
```
