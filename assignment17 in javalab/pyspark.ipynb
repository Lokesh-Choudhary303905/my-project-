```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, desc, sum, avg
spark=SparkSession.builder.appName("SparkCompleteNotes").getOrCreate()
data=spark.read.csv("data.csv",header=True,inferSchema=True)
data.printSchema()
data.show()
```

```python
# Sales descending order
# data.orderBy(desc("sales")).show()
data = data.toDF(*[c.strip() for c in data.columns])
data.createOrReplaceTempView("data")
d=spark.sql("select sales from data order by sales desc")
d.show()
d.write.mode("overwrite").csv("output/salesdesc")
```

```python
# Top 3 products

l=spark.sql("select * from data order by sales desc limit 3")
l.show()
l.write.mode("overwrite").csv("output/top3")
```

```python
# Sales > 80000
f=spark.sql("select * from data where sales>80000")
f.show()
f.write.mode("overwrite").csv("output/filtered")
```
