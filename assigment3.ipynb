```python
def fuc(a):
    max=a[0]
    for i in range(len(a)):
      if (max<a[i]):
        max=a[i]
    return max
a=[2,4,5]
print(fuc(a))     
```

```python
def dis(a):
    b=[]
    for i in a:
        if i not in b:
            b.append(i)
    return b        
    
a=[3,1,5,5,7,7]
print(dis(a))    
```

```python
def mul(a):
    b=1
    for i in a:
        b=b*i
    return b
a=[3,6,1,2]
print(mul(a))        
        
```

```python
def fac(a):
    if(a==1 or a==0):
        return 1
    return a*fac(a-1)
a=5
print(fac(a))    
```

```python
def check_range(num, start, end):
    if num >= start and num <= end:
        print("Number is within the range")
    else:
        print("Number is outside the range")

# Example
check_range(7, 1, 10)
```

```python
def even_numbers(lst):
    for i in lst:
        if i % 2 == 0:
            print(i)

# Example
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
even_numbers(numbers)
```

```python
def prime(num):
    if num <= 1:
        print("Not Prime")
    else:
        for i in range(2, num):
            if num % i == 0:
                print("Not Prime")
                break
        else:
            print("Prime")

# Example
prime(7)
```

```python
def count_case(text):
    upper = 0
    lower = 0

    for i in text:
        if i.isupper():
            upper += 1
        elif i.islower():
            lower += 1

    print("Uppercase letters:", upper)
    print("Lowercase letters:", lower)

# Example
count_case("Hello World")
```
