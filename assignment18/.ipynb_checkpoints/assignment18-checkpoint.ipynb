```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col,udf
from pyspark.sql.types import StringType,IntegerType
import time  
```

```python
spark=SparkSession.builder.appName("assign").getOrCreate()
```

```python
df=spark.range(0,5000000)
df.take(12)
```

```python
print("before part",df.rdd.getNumPartitions())
```

```python
df_repartition=df.repartition(12)
```

```python
print("after partition",df_repartition.rdd.getNumPartitions())
```

```python
print(df_repartition)
```

```python
df_coalesce=df_repartition.coalesce(3)
```

```python
print("Partition after coalesce(3):",df_coalesce.rdd.getNumPartitions())
df_coalesce.show(10)
```

```python
spark.stop()
```
