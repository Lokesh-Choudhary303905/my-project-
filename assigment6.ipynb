```python
#1)
import requests

city = input("Enter city name: ")

api_key = "YOUR_API_KEY"

url=f"https://api.openweathermap.org/data/2.5/weather?q={city}&appid=811013a477bd1ea4de5aa66d89981035&units=metric"

data = requests.get(url)

weather = data.json()

print("City:", weather["name"])
print("Temperature:", weather["main"]["temp"], "°C")
print("Feels Like:", weather["main"]["feels_like"], "°C")
print("Humidity:", weather["main"]["humidity"], "%")
print("Pressure:", weather["main"]["pressure"])
print("Weather:", weather["weather"][0]["description"])
print("Wind Speed:", weather["wind"]["speed"])
print("Country:", weather["sys"]["country"])
```

```python
import random

num = random.randint(1, 100)

while True:
    
    guess = int(input("Guess Number (1-100): "))

    if guess > num:
        print("Too High")

    elif guess < num:
        print("Too Low")

    else:
        print("Correct Number")
        break
```

```python
#more free apis
#1)random joke api
import requests

url = "https://official-joke-api.appspot.com/random_joke"

data = requests.get(url)

joke = data.json()

print("Setup:", joke["setup"])
print("Punchline:", joke["punchline"])
```

```python
#2)random user api
import requests

url = "https://randomuser.me/api/"

data = requests.get(url)

user = data.json()

info = user["results"][0]

print("Name:", info["name"]["first"], info["name"]["last"])
print("Email:", info["email"])
print("Country:", info["location"]["country"])
```
