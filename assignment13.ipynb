```python
import numpy as np
arr1=np.array([[6, -8, 73, -110], [np.nan, -8, 0, 94]])
a=np.nan_to_num(arr1,nan=0)
print(a)
d=np.transpose(a)
print(d)
```

```python
a=np.arange(1,19).reshape(2,3,3)
print(a)
f=np.moveaxis(a,0,2)
print(f)
print(f.shape)
g=np.moveaxis(a,1,2)
print(g,g.shape)
```

```python
arr1=np.array([[6, -8, 73, -110], [np.nan, -8, 0, 94]])
clmean=np.nanmean(arr1,axis=0)

print(clmean)
inds=np.where(np.isnan(arr1))
arr1[inds]=np.take(clmean,inds[1])
print(arr1)
arr1[arr1<0]=0
print(arr1)
```

```python
from scipy import stats
arr1 = np.array([[3, 4],[4,6]]) 
arr2 = np.array([[1, 0],[4,7]]) 
avg = (arr1 + arr2) / 2
print("Average of NumPy arrays:\n", avg)
m = np.concatenate((arr1.flatten(), arr2.flatten()))
print(m)
mean = np.mean(m)
print(mean)
print(np.median(m))

print("mode=",stats.mode(m,keepdims=True).mode[0])

```

```python
a=np.array([[1,-2,3],[-1,3,-1],[2,-5,5]])
b=np.array([[9],[-6],[17]])
a_in=np.linalg.inv(a)
x=np.dot(a_in,b)
print(x)
```

```python
import matplotlib.pyplot as plt

subjects = ['Maths','DSA','Python','DBMS','OS']

sem1 = [75,80,78,70,72]
sem2 = [82,85,88,79,80]

plt.figure(figsize=(8,5))

plt.plot(subjects,
         sem1,
         marker='o',
         label='Semester 1')

plt.plot(subjects,
         sem2,
         marker='s',
         label='Semester 2')

plt.title("Semester Result Comparison")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.grid(True)
plt.legend()

plt.show()
```
